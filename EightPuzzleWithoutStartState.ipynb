```python
import random

GOAL = [
    [8, 7, 6],
    [1, 0, 5],
    [2, 3, 4]
]

# =========================
# Heuristic H(n)
# =========================
def heuristic(state):
    cnt = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != GOAL[i][j]:
                cnt += 1
    return cnt

# =========================
# Tìm ô trống
# =========================
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

# =========================
# Sinh trạng thái con
# =========================
def child_state(state):
    child = []
    x, y = find_blank(state)
    rules = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]
    for dx, dy, action in rules:
        nx = x + dx
        ny = y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [row[:] for row in state]
            new_state[x][y], new_state[nx][ny] = (
                new_state[nx][ny],
                new_state[x][y]
            )
            child.append((new_state, action))
    return child

def print_matrix(state):
    for row in state:
        print(row)
    print()

# =========================
# Sinh START ngẫu nhiên
# =========================
def random_start_state(steps=30):
    state = [row[:] for row in GOAL]
    for _ in range(steps):
        x, y = find_blank(state)
        moves = []
        if x > 0:
            moves.append((x - 1, y))
        if x < 2:
            moves.append((x + 1, y))
        if y > 0:
            moves.append((x, y - 1))
        if y < 2:
            moves.append((x, y + 1))
        nx, ny = random.choice(moves)
        state[x][y], state[nx][ny] = state[nx][ny], state[x][y]
    return state

# =========================
# Greedy Best First Search
# =========================
def Greedy_Search(start):
    FRONTIER = []
    REACHED = []
    FRONTIER.append({
        "state": start,
        "parent": None,
        "action": None,
        "h": heuristic(start)
    })
    while FRONTIER:
        FRONTIER.sort(key=lambda x: x["h"])
        n = FRONTIER.pop(0)
        state = n["state"]
        if state == GOAL:
            print("\nSuccessful!\n")
            path = []
            while n is not None:
                path.append(n)
                n = n["parent"]
            path.reverse()
            return path
        REACHED.append(state)
        for next_state, action in child_state(state):
            in_frontier = any(
                node["state"] == next_state
                for node in FRONTIER
            )
            in_reached = any(
                s == next_state
                for s in REACHED
            )
            if not in_frontier and not in_reached:
                FRONTIER.append({
                    "state": next_state,
                    "parent": n,
                    "action": action,
                    "h": heuristic(next_state)
                })
    return None

# =========================
# MAIN
# =========================
START = random_start_state(30)
print("TRẠNG THÁI BẮT ĐẦU")
print_matrix(START)
result = Greedy_Search(START)
if result:
    print("\n===== ĐƯỜNG ĐI =====\n")
    for step, node in enumerate(result):
        action = node["action"] if node["action"] else "START"
        print(f"Step {step} | Action: {action} | h = {node['h']}")
else:
    print("Không tìm thấy đường đi.")
```
