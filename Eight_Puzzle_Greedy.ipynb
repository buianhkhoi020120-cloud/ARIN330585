```python
GOAL = [
    [1,2,3],
    [4,5,6],
    [7,8,0]
]
START = [
    [1,2,3],
    [4,0,6],
    [7,5,8]
]
# tính H(m)
def heuristic(state):
    cnt = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != GOAL[i][j]:
                cnt += 1
    return cnt
# Tìm ô trống
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
# sinh trạng thái con
def child_state(state):
    child = []
    x, y = find_blank(state)
    # dx, dy, action
    rules = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]
    for dx,dy,action in rules:
        nx = x + dx
        ny = y + dy
        if 0 <= nx <= 2 and 0 <= ny <= 2:
            new_state = [row[:] for row in state]
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            child.append((new_state, action))
    return child
# in ma trận
def print_matrix(state):
    for row in state:
        print(row)
    print()
# thuật toán Greedy
def Greedy_Search(start, goal):
    FRONTIER = []
    REACHED = []
    FRONTIER.append({
        "state": start,
        "parent": None,
        "action": None,
        "h": heuristic(start)
    })
    while FRONTIER:
        # Chọn n có h nhỏ nhất
        FRONTIER.sort(key = lambda x : x["h"])
        n = FRONTIER.pop(0)
        state = n["state"]
        print("Đang xét: ")
        print_matrix(state)
        print("h = ", n["h"])
        print("-" * 30 )
        # Kiểm tra GOAL
        if state == GOAL:
            print("Successful")
            path = []
            while n is not None:
                path.append(n)
                n = n["parent"]
            path.reverse()
            return path
        # thêm vào REACHED
        REACHED.append(state)
        # Sinh các trạng thái con
        child = child_state(state)
        for next_state, action in child:
            # Kiểm tra state đã tồn tại chưa
            in_frontier = any(
                node["state"] == next_state
                for node in FRONTIER
            )
            in_reached = any(
                s == next_state
                for s in REACHED
            )
            if not in_frontier and not in_reached:
                FRONTIER.append({
                    "state": next_state,
                    "parent": n,
                    "action": action,
                    "h": heuristic(next_state)
                })
    return None
# Chạy thuật toán
result = Greedy_Search(START, GOAL)
# In kết quả
if result:
    print("\nĐƯỜNG ĐI\n")
    step = 0
    for node in result:
        print(f"Step: {step}")
        if node["action"]:
            print("Action:", node["action"])
        print("h =", node["h"])
        print_matrix(node["state"])
        step += 1
else:
    print("Không tìm thấy đường đi.")






```
